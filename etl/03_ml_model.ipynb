```python
import pandas as pd

# Load cleaned files
companies = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/companies_clean.csv")
profit = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/profit_clean.csv")
balance = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/balance_clean.csv")
cashflow = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/cashflow_clean.csv")

print("✅ Data loaded")
```

```python
print(companies['id'].head())
print(profit.head())
```

```python
# companies lo id ni company_id ga rename chestham
companies.rename(columns={"id": "company_id"}, inplace=True)
```

```python
companies['company_id'] = companies['company_id'].astype(str).str.strip()
profit['company_id'] = profit['company_id'].astype(str).str.strip()
balance['company_id'] = balance['company_id'].astype(str).str.strip()
cashflow['company_id'] = cashflow['company_id'].astype(str).str.strip()
```

```python
df1 = pd.merge(companies, profit, on="company_id", how="inner")
df2 = pd.merge(df1, balance, on="company_id", how="inner")
final_df = pd.merge(df2, cashflow, on="company_id", how="inner")

print(final_df.shape)
```

```python
final_df.to_csv(
    "D:/Bluestaks/B100 Intelligence/data_clean/final_dataset.csv",
    index=False
)

print("🔥 FINAL DATASET READY")
```

```python
# STEP 1: Final dataset load
import pandas as pd

df = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/final_dataset.csv")

df.head()
```

```python
# STEP 2: Check useful columns
df.columns
```

```python
# Remove duplicate id and year columns
df = df.drop(columns=['id_x', 'id_y', 'id', 'year_x', 'year_y'])

df.columns
```

```python
df[['company_id', 'roe_percentage', 'roce_percentage', 'net_profit']].head()
```

```python
# Group by company and take average

df_grouped = df.groupby('company_id').agg({
    'roe_percentage': 'mean',
    'roce_percentage': 'mean',
    'net_profit': 'mean'
}).reset_index()

df_grouped.head()
```

```python
df_grouped['ml_score'] = (
    df_grouped['roe_percentage'] * 0.4 +
    df_grouped['roce_percentage'] * 0.3 +
    (df_grouped['net_profit'] / 1000) * 0.3
)
```

```python
def classify(score):
    if score > 50:
        return "Strong"
    elif score > 25:
        return "Average"
    else:
        return "Weak"

df_grouped['category'] = df_grouped['ml_score'].apply(classify)
```

```python
df_grouped.to_csv(
    "D:/Bluestaks/B100 Intelligence/data_clean/final_ml_dataset.csv",
    index=False
)

print("🔥 FINAL CLEAN ML DATASET READY")
```
