```python
# STEP 1: Library import
import pandas as pd

# STEP 2: Companies file load
# header=1 because first row title untundi (skip cheyyali)
companies = pd.read_excel(
    "D:/Bluestaks/B100 Intelligence/data_raw/companies.xlsx",
    header=1
)

# STEP 3: First rows check
companies.head()
```

```python
# STEP 4: Column names clean
# strip() → spaces remove
# lower() → lowercase
companies.columns = companies.columns.str.strip().str.lower()

# Check columns
companies.columns
```

```python
# STEP 5: Duplicate rows remove
companies = companies.drop_duplicates()

# STEP 6: Missing values replace
companies = companies.fillna(0)

# STEP 7: Company name clean (remove \n, spaces)
companies['company_name'] = companies['company_name'].str.strip()

# Structure check
companies.info()
```

```python
# STEP 8: Save clean file
companies.to_csv(
    "D:/Bluestaks/B100 Intelligence/data_clean/companies_clean.csv",
    index=False
)

print("✅ Companies cleaned & saved")
```

```python
import pandas as pd
import os

os.makedirs("D:/Bluestaks/B100 Intelligence/data_clean", exist_ok=True)

# Load files
companies = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/companies_clean.csv")
profit = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/profit_clean.csv")
balance = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/balance_clean.csv")
cashflow = pd.read_csv("D:/Bluestaks/B100 Intelligence/data_clean/cashflow_clean.csv")

# 🔥 IMPORTANT FIX
companies['id'] = companies['id'].astype(str)
profit['id'] = profit['id'].astype(str)
balance['id'] = balance['id'].astype(str)
cashflow['id'] = cashflow['id'].astype(str)

# Merge
df1 = pd.merge(companies, profit, on="id", how="inner")
df2 = pd.merge(df1, balance, on="id", how="inner")
final_df = pd.merge(df2, cashflow, on="id", how="inner")

# Save
final_df.to_csv(
    "D:/Bluestaks/B100 Intelligence/data_clean/final_dataset.csv",
    index=False
)

print("🔥 FINAL DATASET CREATED SUCCESSFULLY")
```
